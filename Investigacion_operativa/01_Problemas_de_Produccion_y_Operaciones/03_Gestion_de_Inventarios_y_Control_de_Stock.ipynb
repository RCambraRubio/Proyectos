# Gestión de inventarios y control de stock

## Tipo de problema

Definir el punto de pedido (Reorder Point) y el tamaño de lote (EOQ) considerando demanda estocástica, costes de almacenamiento y penalizaciones por rotura de stock.

## Ejercicio propuesto

Una tienda vende 20 ud/día de media, con desviación típica de 5 ud.  
Abre 250 días/año, pedir cuesta 40 € y almacenar cuesta 2 €/ud/año.  
Plazo fijo: 4 días. Nivel de servicio de ciclo: 95 % (probabilidad de no  
romper stock durante el plazo; no equivale a la tasa de demanda satisfecha).  
Método: lote económico EOQ y punto de pedido con aproximación normal.  
Se suponen demandas diarias independientes, sin descuentos ni restricciones.  
EOQ es óptimo para demanda determinista; añadir seguridad es una política  
aproximada bajo incertidumbre, no un óptimo estocástico demostrado.

## Método elegido para este ejercicio

**Enfoque utilizado:** Fórmulas analíticas.

Se utiliza el lote económico EOQ y una aproximación normal para el stock de seguridad. Se evalúan los dos enteros vecinos del EOQ continuo. El ejemplo fija un nivel de servicio en lugar de optimizar una penalización por rotura.

## Implementación

Ejecuta las celdas de código en orden. Solo se utiliza la biblioteca estándar de Python.

```python
from math import ceil, sqrt
from statistics import NormalDist


def resolver():
    demanda, pedido, mantenimiento = 20 * 250, 40, 2
    continuo = sqrt(2 * demanda * pedido / mantenimiento)
    candidatos = {max(1, int(continuo)), ceil(continuo)}

    def coste(q):
        return demanda * pedido / q + mantenimiento * q / 2

    lote = min(candidatos, key=coste)
    seguridad = NormalDist().inv_cdf(0.95) * 5 * sqrt(4)
    punto = ceil(20 * 4 + seguridad)
    return lote, punto, seguridad, coste(lote) + mantenimiento * seguridad
```

## Resolución y resultados

```python
lote, punto, seguridad, coste = resolver()
print(f'Pedir {lote} unidades cuando la posición de inventario llegue a {punto}.')
print('Posición = existencias + pedidos pendientes - demanda pendiente.')
print(f'Stock de seguridad teórico: {seguridad:.2f} unidades')
print(f'Coste anual aproximado de pedidos y almacenamiento: {coste:.2f} €')
```

## Interpretación

Pedir **447 unidades** cuando la posición de inventario llegue a **97**. El stock de seguridad teórico es 16,45 unidades. El coste anual aproximado es **927,32 €**, sin compras. El lote es óptimo para EOQ determinista; la política completa con incertidumbre es aproximada. El 95 % es servicio de ciclo, no porcentaje de unidades entregadas.

**Alcance:** los datos son sintéticos. Las garantías y supuestos se refieren a esta instancia; al aumentar su tamaño, la enumeración puede resultar costosa.
