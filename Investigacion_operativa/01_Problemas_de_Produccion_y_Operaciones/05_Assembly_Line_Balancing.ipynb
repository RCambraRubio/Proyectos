# Equilibrado de una línea de montaje

## Tipo de problema

Asignar tareas a estaciones de trabajo en una cadena de montaje para igualar los tiempos de ciclo y minimizar el tiempo ocioso de los operarios.

## Formas de resolverlo

Los nombres de la primera columna coinciden con la [tabla del README](../README.md#formas-de-resolverlo).

| Enfoque (README) | Aplicación en este problema | Uso en este ejercicio |
|---|---|---|
| Entera o mixta | Formular asignaciones, capacidades y precedencias con variables enteras. | Alternativa |
| Restricciones | Buscar asignaciones compatibles con tiempos y precedencias. | Alternativa |
| Heurísticas y metaheurísticas | Construir asignaciones mediante prioridades y mejorarlas. | Alternativa |
| Enumeración | Comparar asignaciones hasta encontrar el mínimo número de estaciones. | **Principal** |

**Enfoque utilizado:** Enumeración.

**Técnica concreta:** Probar asignaciones con un número creciente de estaciones y comprobar cargas y precedencias.

## Ejercicio propuesto

Asignar A(2 min), B(3), C(2), D(4) y E(1) a estaciones de montaje.  
Precedencias: A antes de C, B antes de D, C y D antes de E. Ciclo: 5 min.  
Objetivo: mínimo número de estaciones. En una estación las tareas pueden  
seguir su orden de precedencia dentro del mismo ciclo.  
Método exacto: enumerar asignaciones con capacidades y precedencias.

## Método elegido para este ejercicio

**Enfoque utilizado:** Enumeración.

Enumeración exacta: se prueba un número creciente de estaciones y se comprueban tiempos y precedencias. Con ciclo fijo y trabajo total fijo, minimizar estaciones equivale a minimizar el ocio total.

## Implementación

Ejecuta las celdas de código en orden. Solo se utiliza la biblioteca estándar de Python.

```python
from itertools import product


def resolver():
    tareas, tiempos = ('A', 'B', 'C', 'D', 'E'), (2, 3, 2, 4, 1)
    precedencias = ((0, 2), (1, 3), (2, 4), (3, 4))
    for n in range(1, len(tareas) + 1):
        for asignacion in product(range(n), repeat=len(tareas)):
            if any(asignacion[a] > asignacion[b] for a, b in precedencias):
                continue
            cargas = [sum(t for t, e in zip(tiempos, asignacion) if e == k)
                      for k in range(n)]
            if all(0 < c <= 5 for c in cargas):
                estaciones = [[t for t, e in zip(tareas, asignacion) if e == k]
                              for k in range(n)]
                return estaciones, cargas, n * 5 - sum(tiempos)
    raise ValueError('No hay asignación factible')
```

## Resolución y resultados

```python
estaciones, cargas, ocio = resolver()
for i, (tareas, carga) in enumerate(zip(estaciones, cargas), 1):
    print(f'Estación {i}: {", ".join(tareas)} ({carga}/5 min)')
print(f'Mínimo: {len(estaciones)} estaciones; ocio total: {ocio} min/ciclo')
```

## Interpretación

Asignar **A+B**, **C** y **D+E** a tres estaciones. El ocio total es **3 min por ciclo**. Las tareas suman 12 min y cada estación admite 5; hacen falta al menos tres estaciones y la asignación alcanza ese mínimo.

**Alcance:** los datos son sintéticos. Las garantías y supuestos se refieren a esta instancia; al aumentar su tamaño, la enumeración puede resultar costosa.
