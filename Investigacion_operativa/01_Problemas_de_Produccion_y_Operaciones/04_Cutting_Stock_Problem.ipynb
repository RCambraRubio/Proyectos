# Corte de material: Cutting Stock

## Tipo de problema

Minimizar el desperdicio de materia prima (bobinas de papel, planchas de chapa, telas) al cortarla en tamaños específicos solicitados por los clientes.

## Formas de resolverlo

Los nombres de la primera columna coinciden con la [tabla del README](../README.md#formas-de-resolverlo).

| Enfoque (README) | Aplicación en este problema | Uso en este ejercicio |
|---|---|---|
| Entera o mixta | Elegir cantidades enteras de patrones; la generación de columnas evita listarlos todos y puede combinarse con ramificación. | Alternativa |
| Heurísticas y metaheurísticas | Construir rápidamente cortes factibles sin garantía general de óptimo. | Alternativa |
| Enumeración | Generar todos los patrones que caben en una barra de este ejemplo. | **Complementario** |
| Programación dinámica | Encontrar el mínimo de barras para cada demanda pendiente. | **Principal** |

**Enfoque utilizado:** Programación dinámica.

**Enfoque complementario:** Enumeración.

**Técnica concreta:** Enumerar patrones factibles y reutilizar la solución de cada estado de demanda pendiente.

[Formulaciones y generación de columnas](https://scipbook.readthedocs.io/en/latest/bpp.html).

## Ejercicio propuesto

Cortar barras de 10 m para entregar 3 piezas de 6 m y 4 de 4 m.  
Sin sobreproducción; se ignora el grosor de la sierra. Minimizar barras  
minimiza también el desperdicio, porque la longitud demandada es fija.  
Método exacto: generar patrones y programación dinámica sobre demanda pendiente.  
Es una alternativa pequeña y didáctica a la generación de columnas.

## Método elegido para este ejercicio

**Enfoque utilizado:** Programación dinámica.

**Enfoque complementario:** Enumeración.

Se generan todos los patrones que caben en una barra y se aplica programación dinámica a la demanda pendiente. Para este caso pequeño se pueden considerar todos los patrones sin recurrir a generación de columnas.

## Implementación

Ejecuta las celdas de código en orden. Solo se utiliza la biblioteca estándar de Python.

```python
from functools import lru_cache


def resolver():
    patrones = [(a, b) for a in range(2) for b in range(3)
                if 0 < 6 * a + 4 * b <= 10]

    @lru_cache(None)
    def cortar(a, b):
        if (a, b) == (0, 0):
            return 0, ()
        opciones = []
        for x, y in patrones:
            if x <= a and y <= b:
                n, plan = cortar(a - x, b - y)
                opciones.append((1 + n, ((x, y),) + plan))
        return min(opciones)

    n, plan = cortar(3, 4)
    assert sum(a for a, _ in plan) == 3 and sum(b for _, b in plan) == 4
    assert all(6 * a + 4 * b <= 10 for a, b in plan)
    return n, plan, 10 * n - 34
```

## Resolución y resultados

```python
n, plan, residuo = resolver()
for i, (a, b) in enumerate(plan, 1):
    print(f'Barra {i}: {a} piezas de 6 m y {b} piezas de 4 m')
print(f'Mínimo: {n} barras; desperdicio: {residuo} m')
```

## Interpretación

Se necesitan **4 barras** y sobran **6 m**: tres barras con piezas de 6+4 m y otra con una pieza de 4 m. La demanda suma 34 m, de modo que hacen falta al menos cuatro barras de 10 m. La solución alcanza esa cota.

**Alcance:** los datos son sintéticos. Las garantías y supuestos se refieren a esta instancia; al aumentar su tamaño, la enumeración puede resultar costosa.
