# Planificación estratégica de capacidad

## Tipo de problema

Decidir cuándo y cuánto expandir la capacidad productiva de una empresa considerando la incertidumbre en la demanda futura y las decisiones de la competencia.

## Formas de resolverlo

Los nombres de la primera columna coinciden con la [tabla del README](../README.md#formas-de-resolverlo).

| Enfoque (README) | Aplicación en este problema | Uso en este ejercicio |
|---|---|---|
| Entera o mixta | Decidir ampliaciones y fechas con módulos indivisibles. | Alternativa |
| Modelos estocásticos y simulación | Usar un árbol de escenarios con ampliaciones adaptadas a la demanda; aquí no se realiza simulación. | **Principal** |
| Programación dinámica | Relacionar decisiones por periodo con la capacidad instalada. | Alternativa |
| Enumeración | Comparar las pocas ampliaciones iniciales y posteriores posibles. | **Complementario** |

**Enfoque utilizado:** Modelos estocásticos y simulación.

**Enfoque complementario:** Enumeración.

**Técnica concreta:** Árbol de decisión de dos etapas con demanda incierta; se enumeran ampliaciones y se calcula su coste esperado.

## Ejercicio propuesto

Expandir una planta durante dos años con demanda incierta en el segundo.  
Capacidad inicial: 2 ud/año. Cada módulo añade 2 ud/año y cuesta 8 miles €;  
se permiten hasta dos módulos en total. Demanda: año 1 = 3, año 2 = 3 con  
probabilidad 0.6 o 5 con probabilidad 0.4. Toda demanda debe satisfacerse.  
Los módulos entran en servicio al principio del año y no se retiran.  
Objetivo: mínimo coste esperado descontado (factor del año 2: 0.9).  
Método exacto: árbol de decisión de dos etapas con recurso. La decisión inicial  
es común a ambos escenarios; en el año 2 se observa la demanda antes de ampliar.

## Método elegido para este ejercicio

**Enfoque utilizado:** Modelos estocásticos y simulación.

**Enfoque complementario:** Enumeración.

Árbol de decisión de dos etapas: se compara una ampliación inicial común a todos los escenarios y se permite una segunda ampliación después de observar la demanda. Se minimiza el coste esperado descontado. No se modelan competidores en este ejemplo.

## Implementación

Ejecuta las celdas de código en orden. Solo se utiliza la biblioteca estándar de Python.

```python
def resolver():
    escenarios = (('Baja', 0.6, 3), ('Alta', 0.4, 5))
    mejor = float('inf'), None, None
    for inicial in range(3):
        if 2 + 2 * inicial < 3:
            continue
        esperado, politica = 8 * inicial, []
        for nombre, probabilidad, demanda in escenarios:
            posibles = [n for n in range(3 - inicial)
                        if 2 + 2 * (inicial + n) >= demanda]
            if not posibles:
                break
            adicional = min(posibles)
            esperado += 0.9 * probabilidad * 8 * adicional
            politica.append((nombre, adicional, 2 + 2 * (inicial + adicional)))
        else:
            if esperado < mejor[0]:
                mejor = esperado, inicial, politica
    assert mejor[1] is not None
    return mejor
```

## Resolución y resultados

```python
coste, inicial, politica = resolver()
print(f'Año 1: instalar {inicial} módulo(s)')
for escenario, adicional, capacidad in politica:
    print(f'Año 2, demanda {escenario}: añadir {adicional}; capacidad={capacidad}')
print(f'Coste esperado mínimo actualizado: {coste:.2f} miles €')
```

## Interpretación

Instalar **un módulo en el año 1** y añadir otro en el año 2 solo con demanda alta. El coste esperado actualizado es **10,88 miles €**: 8 + 0,9 × 0,4 × 8. La decisión inicial no utiliza información futura.

**Alcance:** los datos son sintéticos. Las garantías y supuestos se refieren a esta instancia; al aumentar su tamaño, la enumeración puede resultar costosa.
