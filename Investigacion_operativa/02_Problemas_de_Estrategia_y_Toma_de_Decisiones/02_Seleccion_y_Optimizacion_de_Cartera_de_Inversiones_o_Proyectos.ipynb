# Selección de inversiones o proyectos

## Tipo de problema

Elegir la combinación óptima de proyectos de I+D o activos financieros que maximicen el retorno esperado manteniendo el riesgo por debajo de un umbral aceptable.

## Formas de resolverlo

Los nombres de la primera columna coinciden con la [tabla del README](../README.md#formas-de-resolverlo).

| Enfoque (README) | Aplicación en este problema | Uso en este ejercicio |
|---|---|---|
| Entera o mixta | Seleccionar proyectos indivisibles con presupuesto e incompatibilidades. | Alternativa |
| No lineal o cuadrática | Modelar una cartera financiera media-varianza de Markowitz. | Alternativa |
| Modelos estocásticos y simulación | Representar beneficios o costes inciertos mediante escenarios. | Alternativa |
| Enumeración | Comparar todas las combinaciones binarias de los cinco proyectos. | **Principal** |

**Enfoque utilizado:** Enumeración.

**Técnica concreta:** Comparar decisiones binarias de proyectos respetando presupuesto, riesgo y dependencias.

## Ejercicio propuesto

Seleccionar proyectos de una empresa con presupuesto de 10 miles de euros.  
Proyecto:             A    B    C    D    E  
Inversión (miles €):   4    5    3    6    2  
Beneficio (miles €):   7    9    6   11    3  
Puntos de riesgo:     2    4    3    4    1  
Riesgo total <= 7; C requiere A y B es incompatible con D.  
Objetivo: maximizar beneficio. Método exacto: enumeración de decisiones binarias.  
El riesgo aditivo es una puntuación didáctica, no una varianza financiera.

## Método elegido para este ejercicio

**Enfoque utilizado:** Enumeración.

Para la variante de proyectos se enumeran decisiones binarias y se filtran presupuesto, riesgo, dependencia e incompatibilidad. El riesgo aquí es una puntuación aditiva: no se usa la varianza de una cartera financiera ni el modelo de Markowitz.

## Implementación

Ejecuta las celdas de código en orden. Solo se utiliza la biblioteca estándar de Python.

```python
from itertools import product


def resolver():
    inversion, beneficio, riesgo = (4, 5, 3, 6, 2), (7, 9, 6, 11, 3), (2, 4, 3, 4, 1)
    mejor = (-1, (), 0, 0)
    for x in product((0, 1), repeat=5):
        coste = sum(a * b for a, b in zip(x, inversion))
        exposicion = sum(a * b for a, b in zip(x, riesgo))
        if coste > 10 or exposicion > 7 or x[2] > x[0] or x[1] + x[3] > 1:
            continue
        valor = sum(a * b for a, b in zip(x, beneficio))
        if valor > mejor[0]:
            mejor = valor, tuple(n for n, elegido in zip('ABCDE', x) if elegido), coste, exposicion
    return mejor
```

## Resolución y resultados

```python
beneficio, proyectos, inversion, riesgo = resolver()
print(f'Proyectos: {", ".join(proyectos)}')
print(f'Beneficio máximo: {beneficio} miles €; inversión: {inversion} miles €')
print(f'Riesgo: {riesgo}/7 puntos')
```

## Interpretación

Elegir **A y D**: inversión de **10 miles €**, beneficio de **18 miles €** y riesgo de **6 puntos**, por debajo del máximo de 7. Se comparan todas las combinaciones factibles.

**Alcance:** los datos son sintéticos. Las garantías y supuestos se refieren a esta instancia; al aumentar su tamaño, la enumeración puede resultar costosa.
