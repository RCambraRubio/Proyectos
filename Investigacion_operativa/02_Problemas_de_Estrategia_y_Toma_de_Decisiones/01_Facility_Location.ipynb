# Localización de instalaciones

## Tipo de problema

Decidir dónde ubicar nuevas plantas, centros de distribución o tiendas para minimizar costes de transporte y maximizar la cobertura de mercado.

## Formas de resolverlo

Los nombres de la primera columna coinciden con la [tabla del README](../README.md#formas-de-resolverlo).

| Enfoque (README) | Aplicación en este problema | Uso en este ejercicio |
|---|---|---|
| Entera o mixta | Modelar aperturas y asignaciones; p-mediana y cobertura son variantes de formulación. | Alternativa |
| Heurísticas y metaheurísticas | Mejorar ubicaciones mediante aperturas, cierres e intercambios. | Alternativa |
| Enumeración | Comparar todos los pares de almacenes candidatos de esta instancia. | **Principal** |

**Enfoque utilizado:** Enumeración.

**Técnica concreta:** Comparar pares de almacenes y asignar cada cliente al abierto con menor coste, sin límites de capacidad.

[Formulaciones de localización de instalaciones](https://scipbook.readthedocs.io/en/latest/flp.html).

## Ejercicio propuesto

Elegir exactamente 2 almacenes entre Norte, Centro y Sur.  
Cuatro clientes demandan 2, 3, 2 y 1 unidades. La matriz indica coste de  
transporte por unidad; abrir cuesta 8, 6 y 5 €. Capacidad ilimitada.  
Objetivo: apertura más transporte; cada cliente usa un solo almacén abierto.  
Método exacto: enumerar pares y asignar cada cliente al más barato del par.  
Es localización con costes fijos y número de centros prescrito.

## Método elegido para este ejercicio

**Enfoque utilizado:** Enumeración.

Se enumeran los pares de almacenes. Para cada par, cada cliente se asigna al almacén abierto con menor coste unitario. Esto es válido porque las capacidades son ilimitadas. Se suman apertura y transporte.

## Implementación

Ejecuta las celdas de código en orden. Solo se utiliza la biblioteca estándar de Python.

```python
from itertools import combinations


def resolver():
    nombres = ('Norte', 'Centro', 'Sur')
    fijos, demanda = (8, 6, 5), (2, 3, 2, 1)
    costes = ((1, 4, 7, 8), (4, 2, 3, 5), (8, 6, 2, 1))
    mejor = float('inf'), (), ()
    for abiertos in combinations(range(3), 2):
        asignacion = tuple(min(abiertos, key=lambda i: costes[i][j])
                           for j in range(4))
        total = sum(fijos[i] for i in abiertos) + sum(
            demanda[j] * costes[i][j] for j, i in enumerate(asignacion))
        if total < mejor[0]:
            mejor = total, abiertos, asignacion
    total, abiertos, asignacion = mejor
    assert len(abiertos) == 2 and all(i in abiertos for i in asignacion)
    return total, tuple(nombres[i] for i in abiertos), tuple(nombres[i] for i in asignacion)
```

## Resolución y resultados

```python
total, abiertos, asignacion = resolver()
print(f'Abrir: {", ".join(abiertos)}; coste mínimo: {total} €')
for cliente, almacen in enumerate(asignacion, 1):
    print(f'Cliente {cliente} atendido desde {almacen}')
```

## Interpretación

Abrir **Centro y Sur**. Centro sirve a C1 y C2; Sur, a C3 y C4. El coste mínimo es **30 €**, incluyendo 11 € de apertura y 19 € de transporte. El ejemplo minimiza costes con exactamente dos centros.

**Alcance:** los datos son sintéticos. Las garantías y supuestos se refieren a esta instancia; al aumentar su tamaño, la enumeración puede resultar costosa.
