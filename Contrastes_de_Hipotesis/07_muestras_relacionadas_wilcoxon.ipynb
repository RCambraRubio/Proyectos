El **Wilcoxon de rangos con signo** evalúa el centro de los cambios pareados sin exigir normalidad. Usamos **D = después − antes**.

## 1. Supuestos

| Supuesto | Qué significa |
|---|---|
| **Parejas definidas** | Antes y después corresponden a la misma persona. |
| **Independencia entre personas** | Los cambios de distintas personas son independientes. |
| **Diferencias cuantitativas** | Tiene sentido comparar y ordenar sus magnitudes. |
| **Simetría de las diferencias** | Los cambios se distribuyen simétricamente alrededor de su centro. |

No requiere normalidad, pero **sí simetría** para esta interpretación. Los ceros y empates afectan al p-valor.

## 2. Ejemplo: puntuación antes y después

Simulamos **20 personas independientes**. Los cambios siguen una distribución uniforme entre −2 y 8: **no normal**, pero simétrica alrededor de 3. Creamos después = antes + cambio.

- **H₀:** las diferencias son simétricas alrededor de cero.
- **H₁:** su centro es distinto de cero (bilateral). **α = 0,05.**

Requiere NumPy y SciPy; no necesita archivos de datos.

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(42)
antes = rng.normal(60, 8, size=20)
despues = antes + rng.uniform(-2, 8, size=20)
d = despues - antes

# Datos continuos simulados: sin ceros ni empates en las magnitudes.
assert (d != 0).all() and len(np.unique(np.abs(d))) == len(d)
resultado = stats.wilcoxon(d, alternative='two-sided', method='exact')
p = resultado.pvalue
print('Primeros cambios:', np.round(d[:5], 2))
print(f'Mediana del cambio: {np.median(d):.2f} puntos')
print(f'W = {resultado.statistic:.1f}; p = {p:.3g}')
print('Rechazamos H₀.' if p < 0.05 else 'No rechazamos H₀.')
```

## 3. Interpretación

La mediana observada del cambio es **2.68 puntos**. Como **p = 0.000708 < 0,05**, hay evidencia de un cambio central positivo.

Bajo el supuesto de simetría, se interpreta como una mediana del cambio distinta de cero. No es un contraste del cambio **medio**.
