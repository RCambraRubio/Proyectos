# Muestras independientes: t de Student

[Fundamentos](00_fundamentos.ipynb) · [t pareada](01_muestras_relacionadas_t_pareada.ipynb)

La **t de Student para muestras independientes** compara las **medias de dos grupos distintos**, suponiendo varianzas poblacionales iguales.

## 1. Supuestos

| Supuesto | Qué significa |
|---|---|
| **Datos cuantitativos comparables** | Ambos grupos miden la misma variable y usan la misma escala. |
| **Observaciones independientes** | No hay parejas ni dependencia dentro de los grupos o entre ellos. |
| **Normalidad por grupo** | Con muestras pequeñas, cada población debe ser aproximadamente normal. Con muestras grandes puede bastar una aproximación por TCL, con varianza finita y sin extremos dominantes. |
| **Varianzas iguales** | Las dos poblaciones tienen la misma dispersión; las varianzas muestrales no tienen que coincidir exactamente. |

Los tamaños de los grupos pueden diferir. Si no puedes justificar varianzas iguales, utiliza **Welch** (`equal_var=False`).


## 2. Ejemplo: puntuaciones de dos grupos

Simulamos **30 personas en A y 30 personas distintas en B**:

- Poblaciones normales con medias 60 y 66, y la **misma desviación típica: 8**.
- Observaciones generadas de forma independiente; los supuestos se cumplen por construcción.
- **H₀:** $\mu_B - \mu_A = 0$. **H₁:** $\mu_B - \mu_A \ne 0$ (bilateral).
- **α = 0,05.** Calculamos **B − A**; una diferencia positiva favorece a B.

Requiere NumPy y SciPy; no necesita archivos de datos.


```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(42)
a = rng.normal(loc=60, scale=8, size=30)
b = rng.normal(loc=66, scale=8, size=30)

# equal_var=True: Student con varianza común.
resultado = stats.ttest_ind(b, a, equal_var=True, alternative='two-sided')
ic = resultado.confidence_interval(confidence_level=0.95)
print(f'Media A: {a.mean():.2f}; media B: {b.mean():.2f}')
print(f'Diferencia B − A: {b.mean() - a.mean():.2f} puntos')
print(f'IC 95 %: [{ic.low:.2f}, {ic.high:.2f}] puntos')
print(f't = {resultado.statistic:.2f}; p = {resultado.pvalue:.4f}')
print('Rechazamos H₀.' if resultado.pvalue < 0.05 else 'No rechazamos H₀.')
```

## 3. Interpretación

El grupo B obtuvo **6.78 puntos más de media** que A (IC 95 %: **3.51 a 10.05**). Como **p = 0.0001 < 0,05**, rechazamos H₀: hay evidencia de una diferencia de medias, bajo los supuestos.

Son datos simulados: el resultado ilustra el método.
