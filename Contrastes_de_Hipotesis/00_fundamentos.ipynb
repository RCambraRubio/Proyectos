# Fundamentos para elegir un contraste

## 1. Empieza por la pregunta

Un **contraste de hipótesis** evalúa una afirmación sobre una población usando una muestra. El test se elige según lo que quieras comparar:

| Objetivo | Qué analizas |
|---|---|
| Comparar promedios | Medias |
| Comparar valores centrales al ordenar los datos | Medianas |
| Detectar si un grupo tiende a tener valores mayores | Rangos |
| Evaluar cómo varían juntas dos variables | Asociación o correlación |
| Comparar frecuencias relativas de un resultado | Proporciones |

Identifica también:

- **Variable:** cuantitativa, ordinal (categorías ordenadas) o categórica.
- **Diseño:** una muestra, grupos independientes o mediciones relacionadas.
- **Independencia:** ausencia de dependencia entre observaciones; en datos pareados, se exige entre parejas.

## 2. Hipótesis y resultado

Fija antes del análisis:

| Concepto | Significado | Ejemplo: dos medias |
|---|---|---|
| **H₀** | Afirmación que se pone a prueba | $\mu_A = \mu_B$ |
| **H₁** | Alternativa bilateral o unilateral | $\mu_A \ne \mu_B$ o $\mu_A > \mu_B$ |
| **α** (nivel de significación) | Límite de probabilidad de rechazar H₀ siendo cierta, bajo los supuestos | Habitualmente 0,05 |

**p-valor:** probabilidad, bajo H₀ y el modelo, de obtener un estadístico tan extremo como el observado o más. **No es la probabilidad de H₀.**

| Resultado | Interpretación |
|---|---|
| **p < α** | Se rechaza H₀. |
| **p ≥ α** | No se rechaza H₀; no demuestra igualdad. |

- **Significación ≠ importancia práctica:** informa del tamaño del efecto y su intervalo de confianza.
- **Varios contrastes:** considera ajustes por comparaciones múltiples.
- **Test global significativo:** no identifica por sí solo qué grupos difieren.

## 3. Datos, media y normalidad

| Concepto | Significado |
|---|---|
| **Media poblacional $\mu$** | Valor fijo, generalmente desconocido. |
| **Media muestral $\bar X$** | Estimación que cambia de una muestra a otra. |
| **Error estándar** | Variabilidad de una estimación entre muestras; para la media, se estima con $s/\sqrt{n}$. |

**Distribución de la media:**

- **Población normal + observaciones iid:** la media es normal para cualquier tamaño muestral.
- **Población no normal:** el **TCL**, Teorema Central del Límite, aproxima la media estandarizada a una normal al aumentar $n$. Su versión clásica requiere observaciones iid y varianza finita positiva.
- **iid:** independientes e idénticamente distribuidas. En dos grupos, se aplica dentro de cada grupo; en datos pareados, a las diferencias entre parejas.

El TCL **no convierte los datos en normales**. No existe un tamaño universal suficiente como «n ≥ 30».

## 4. Test de normalidad

**Shapiro–Wilk:** H₀ = la muestra procede de una población normal; presupone observaciones independientes.

| Resultado | Interpretación |
|---|---|
| **p < α** | Evidencia contra la normalidad. |
| **p ≥ α** | No se detecta una desviación; no prueba normalidad. |

- **Qué revisar:** cada grupo por separado, o las **diferencias** en datos pareados.
- **Limitaciones:** con pocos datos puede no detectar desviaciones; con muchos, detectar desviaciones mínimas.
- **Complemento:** histograma y gráfico Q–Q (cuantiles observados frente a los de una normal).

## 5. Tipos de tests

Un **parámetro** describe una población: media $\mu$, varianza $\sigma^2$, proporción $p$, etc.

| Tipo | En qué se basa | Ejemplos |
|---|---|---|
| **Paramétrico** | Una familia de distribuciones definida por un número fijo de parámetros. | t: modelo normal clásico; binomial: modelo de éxitos y fracasos. |
| **No paramétrico** | No fija una familia paramétrica para la distribución poblacional; suele usar rangos o signos. | Mann–Whitney, Wilcoxon, test de los signos. |

- **Paramétrico ≠ normal:** también hay modelos binomiales, Poisson, etc.
- **No paramétrico ≠ sin supuestos:** puede exigir independencia, simetría u otras condiciones.
- **Rangos:** 10, 20 y 100 pasan a 1, 2 y 3; conservan el orden, no las distancias.
- **Rangos ≠ medianas:** su interpretación como medianas requiere condiciones adicionales, indicadas abajo.

### Tests habituales

Condiciones comunes: muestreo adecuado e independencia entre observaciones, o entre sujetos/parejas en diseños relacionados. 

| Test | Tipo | Uso | Condiciones principales |
|---|---|---|---|
| **t de una muestra** | Paramétrico | Media frente a un valor. | Datos cuantitativos; normalidad o aproximación suficiente por TCL. |
| **[t de Student independiente](03_muestras_independientes_t_student.ipynb)** | Paramétrico | Medias de dos grupos. | Grupos independientes; normalidad por grupo o aproximación adecuada; **varianzas iguales**. |
| **[t de Welch](02_muestras_independientes_t_welch.ipynb)** | Paramétrico | Medias de dos grupos. | Grupos independientes; normalidad por grupo o aproximación adecuada; permite **varianzas distintas**. |
| **[t pareada](01_muestras_relacionadas_t_pareada.ipynb)** | Paramétrico | Cambio medio entre dos mediciones. | Parejas definidas; **diferencias** normales o aproximación adecuada de su media. |
| **[ANOVA de un factor](05_grupos_independientes_anova.ipynb)** | Paramétrico | Medias de varios grupos independientes. | Datos cuantitativos; errores normales por grupo; varianzas iguales. |
| **ANOVA de Welch** | Paramétrico | Medias de varios grupos independientes. | Datos cuantitativos; normalidad por grupo o aproximación adecuada; permite varianzas distintas. |
| **ANOVA de medidas repetidas** | Paramétrico | Medias en ≥ 3 condiciones de los mismos sujetos. | Normalidad multivariante de los errores; esfericidad (igual varianza de las diferencias entre condiciones) o corrección. |
| **[Wilcoxon pareado](07_muestras_relacionadas_wilcoxon.ipynb)** | No paramétrico | Centro de las diferencias frente a cero. | Diferencias cuantitativas y **simétricas**; tratar ceros y empates. Bajo simetría, contrasta la mediana del cambio. |
| **Test de los signos** | No paramétrico | Mediana frente a un valor o signo de los cambios. | Datos comparables con la referencia; sin simetría. Interpretación simple como mediana: continuidad; tratar ceros. |
| **[Mann–Whitney](06_grupos_independientes_mann_whitney.ipynb)** | No paramétrico | Rangos de dos grupos independientes. | Datos al menos ordinales; H₀ habitual: distribuciones iguales. Interpretación como medianas: misma forma y dispersión salvo desplazamiento. |
| **Kruskal–Wallis** | No paramétrico | Rangos de varios grupos independientes. | Datos al menos ordinales; tamaños suficientes para aproximación chi-cuadrado. Interpretación como medianas: misma forma y dispersión. |
| **Friedman** | No paramétrico | Rangos en ≥ 3 condiciones relacionadas. | Bloques completos e independientes; condiciones intercambiables bajo H₀. Con pocos datos, valorar cálculo exacto o permutación. |
| **Pearson** | Paramétrico clásico | Correlación lineal frente a cero. | Variables cuantitativas; normalidad bivariante para el test exacto; sensible a extremos. |
| **Spearman / Kendall** | No paramétrico | Asociación por rangos, especialmente monótona. | Datos al menos ordinales; tratar empates; p-valor adecuado al tamaño muestral. |
| **[Chi-cuadrado de independencia](04_variables_categoricas_chi_cuadrado.ipynb)** | No paramétrico* | Asociación entre variables categóricas. | Recuentos; categorías excluyentes; frecuencias esperadas suficientes. Guía habitual: ninguna < 1 y al menos 80 % ≥ 5. |
| **Fisher exacto (2 × 2)** | No paramétrico* | Asociación entre dos variables binarias. | Recuentos independientes; cálculo condicionado a los márgenes; admite frecuencias pequeñas. |
| **Binomial exacto** | Paramétrico | Proporción frente a un valor. | Número fijo de ensayos independientes; dos resultados y probabilidad de éxito constante. |

*Clasificación convencional para chi-cuadrado y Fisher; ambos usan modelos de recuentos. **Exacto** se refiere al cálculo del p-valor.

**Notas:** las t requieren datos cuantitativos; las aproximaciones de muestras grandes necesitan varianza finita positiva. Welch usa una aproximación incluso bajo normalidad. En los tests de rangos, los empates pueden requerir ajustes del p-valor.

## 6. Ver el TCL en una simulación

- **Población:** exponencial de escala 1; asimétrica, con $\mu = 1$ y $\sigma = 1$.
- **Primer panel:** 20.000 observaciones individuales.
- **Otros paneles:** 5.000 muestras para cada tamaño $n = 5, 30, 150$.
- **Estandarización:** restar la media poblacional y dividir por el error estándar:

$$Z_n = \frac{\bar X_n - \mu}{\sigma/\sqrt{n}} = (\bar X_n - 1)\sqrt{n}.$$

**Qué observar:** las medias estandarizadas se aproximan a la normal estándar (curva negra) al aumentar $n$; los datos individuales siguen siendo asimétricos.

Ejecuta las dos celdas en orden. Requieren NumPy, Matplotlib y SciPy; no necesitan datasets externos.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
rng_tcl = np.random.default_rng(2718)
fig, axes = plt.subplots(1, 4, figsize=(14, 3.4))
poblacion_demo = rng_tcl.exponential(scale=1, size=20000)
axes[0].hist(poblacion_demo, bins=60, density=True, color='steelblue')
axes[0].set(title='Datos individuales: asimétricos', xlabel='Valor', ylabel='Densidad', xlim=(0, 7))
z_grid = np.linspace(-4, 5, 300)
for ax, tam in zip(axes[1:], [5, 30, 150]):
    medias_demo = rng_tcl.exponential(scale=1, size=(5000, tam)).mean(axis=1)
    z_demo = (medias_demo - 1) * np.sqrt(tam)
    ax.hist(z_demo, bins=45, density=True, alpha=.7)
    ax.plot(z_grid, stats.norm.pdf(z_grid), color='black', label='Normal estándar')
    ax.set(title=f'Medias estandarizadas, n={tam}', xlabel='Media estandarizada', xlim=(-4, 5))
axes[-1].legend(fontsize=8)
plt.tight_layout()
plt.show()
```
