# Grupos independientes: U de Mann–Whitney

[Fundamentos](00_fundamentos.ipynb)

La **U de Mann–Whitney** compara dos grupos mediante **rangos**, sin exigir normalidad.

## 1. Supuestos

| Supuesto | Qué significa |
|---|---|
| **Grupos independientes** | Personas distintas; sin dependencia dentro ni entre grupos. |
| **Datos ordenables** | Variable al menos ordinal, con la misma escala en ambos grupos. |
| **Empates tratados** | Los valores repetidos requieren un cálculo del p-valor adecuado. |

Para interpretar el resultado como diferencia de **medianas**, las distribuciones deben tener la misma forma y dispersión, salvo desplazamiento.

## 2. Ejemplo: tiempos de espera

Simulamos **40 clientes distintos por servicio**. Los tiempos son exponenciales (asimétricos); B tiene la misma distribución desplazada 3 minutos. No se supone normalidad.

- **H₀:** las distribuciones de A y B son iguales.
- **H₁:** un grupo tiende a presentar tiempos mayores. Contraste **bilateral**, **α = 0,05.**

Requiere NumPy y SciPy; no necesita archivos de datos.

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(42)
a = rng.exponential(scale=3, size=40)
b = rng.exponential(scale=3, size=40) + 3

resultado = stats.mannwhitneyu(b, a, alternative='two-sided', method='asymptotic')
p = resultado.pvalue
superioridad = resultado.statistic / (len(a) * len(b))
print(f'Mediana A: {np.median(a):.2f}; mediana B: {np.median(b):.2f} minutos')
print(f'U de B = {resultado.statistic:.1f}; p = {p:.3g}')
print(f'Proporción de comparaciones B > A (medio peso a empates): {superioridad:.1%}')
print('Rechazamos H₀.' if p < 0.05 else 'No rechazamos H₀.')
```

## 3. Interpretación

Como **p = 2.9e-08 < 0,05**, hay evidencia contra la igualdad de distribuciones. B presenta tiempos mayores en el **86.1%** de las comparaciones entre observaciones.

En este ejemplo, la misma forma y dispersión permiten interpretar el resultado como una diferencia de localización, incluida la mediana.
