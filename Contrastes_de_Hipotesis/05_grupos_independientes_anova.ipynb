# ANOVA: varios grupos independientes

[Fundamentos](00_fundamentos.ipynb)

El **ANOVA de un factor** compara las medias de varios grupos. Aquí usamos **cuatro**; también se utiliza con tres.

## 1. Supuestos

| Supuesto | Qué significa |
|---|---|
| **Variable cuantitativa** | Misma medida y escala en todos los grupos. |
| **Independencia** | Observaciones independientes dentro y entre grupos. |
| **Normalidad por grupo** | Errores aproximadamente normales dentro de cada grupo. |
| **Varianzas iguales** | Dispersión poblacional común a todos los grupos. |

Si las varianzas difieren, considera ANOVA de Welch.

## 2. Ejemplo: cuatro métodos de aprendizaje

Simulamos **25 personas distintas por método**, con puntuaciones normales, medias 60, 64, 68 y 72 y desviación típica común de 8. Los supuestos se cumplen por construcción.

- **H₀:** las cuatro medias poblacionales son iguales.
- **H₁:** al menos una difiere. **α = 0,05.**

Requiere NumPy y SciPy; no necesita archivos de datos.

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(42)
grupos = [rng.normal(media, 8, size=25) for media in [60, 64, 68, 72]]
for nombre, datos in zip('ABCD', grupos):
    print(f'Media {nombre}: {datos.mean():.2f}')

resultado = stats.f_oneway(*grupos)  # ANOVA clásico: varianzas iguales.
p = resultado.pvalue
print(f'F = {resultado.statistic:.2f}; p = {p:.3g}')
print('Rechazamos H₀.' if p < 0.05 else 'No rechazamos H₀.')
```

## 3. Interpretación

Como **p = 7.47e-07 < 0,05**, hay evidencia de que **no todas las medias son iguales**.

El test no identifica qué grupos difieren. Para ello se necesitan comparaciones posteriores, por ejemplo **Tukey**, con control de comparaciones múltiples.
