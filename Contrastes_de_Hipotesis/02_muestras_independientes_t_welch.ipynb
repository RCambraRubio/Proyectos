# Muestras independientes: t de Welch

[Fundamentos](00_fundamentos.ipynb) · [t de Student](03_muestras_independientes_t_student.ipynb)

La **t de Welch** compara las **medias de dos grupos independientes** sin exigir varianzas iguales.

## 1. Supuestos

| Supuesto | Qué significa |
|---|---|
| **Datos cuantitativos comparables** | Misma variable y escala en ambos grupos. |
| **Independencia** | Observaciones independientes dentro y entre grupos; sin emparejamiento. |
| **Normalidad o aproximación suficiente** | Poblaciones aproximadamente normales, o muestras suficientemente grandes en ambos grupos para el TCL, con varianza finita y sin extremos dominantes. |

Permite **varianzas y tamaños distintos**. Usa una aproximación a la distribución t; no exige un test previo de igualdad de varianzas.


## 2. Ejemplo: puntuaciones de dos grupos

Simulamos personas distintas e independientes:

- **A:** 30 personas; población normal con media 60 y desviación típica 5.
- **B:** 45 personas; población normal con media 67 y desviación típica 12.
- **H₀:** $\mu_B - \mu_A = 0$. **H₁:** $\mu_B - \mu_A \ne 0$ (bilateral).
- **α = 0,05.** Calculamos **B − A**. Las varianzas son diferentes por construcción.

Requiere NumPy y SciPy; no necesita archivos de datos.


```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(42)
a = rng.normal(loc=60, scale=5, size=30)
b = rng.normal(loc=67, scale=12, size=45)

# equal_var=False: Welch, sin suponer varianzas iguales.
resultado = stats.ttest_ind(b, a, equal_var=False, alternative='two-sided')
ic = resultado.confidence_interval(confidence_level=0.95)
print(f'Media A: {a.mean():.2f}; media B: {b.mean():.2f}')
print(f'Diferencia B − A: {b.mean() - a.mean():.2f} puntos')
print(f'IC 95 %: [{ic.low:.2f}, {ic.high:.2f}] puntos')
print(f't = {resultado.statistic:.2f}; gl aproximados = {resultado.df:.2f}; p = {resultado.pvalue:.3g}')
print('Rechazamos H₀.' if resultado.pvalue < 0.05 else 'No rechazamos H₀.')
```

## 3. Interpretación

B obtuvo **7.13 puntos más de media** que A (IC 95 %: **3.91 a 10.35**). Como **p = 3.89e-05 < 0,05**, hay evidencia de una diferencia de medias, bajo los supuestos.

Son datos simulados: el resultado ilustra el método.
