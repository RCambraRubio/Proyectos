# Chi-cuadrado: variables categóricas

[Fundamentos](00_fundamentos.ipynb)

El **chi-cuadrado de independencia** evalúa si dos variables categóricas están asociadas.

## 1. Supuestos

| Supuesto | Qué significa |
|---|---|
| **Recuentos** | La tabla contiene personas o casos, no porcentajes. |
| **Independencia** | Cada persona aporta una sola observación. |
| **Categorías excluyentes** | Cada caso pertenece a una única celda. |
| **Frecuencias esperadas suficientes** | Como criterio sencillo, todas ≥ 5; se calculan bajo H₀. |

No requiere normalidad. En tablas 2 × 2 con frecuencias pequeñas, considera Fisher exacto.

## 2. Ejemplo: dispositivo y compra

Datos ficticios de **200 clientes independientes**, uno por registro:

| Dispositivo | Compra | No compra |
|---|---|---|
| Móvil | 30 | 70 |
| Ordenador | 50 | 50 |

- **H₀:** dispositivo y compra son independientes.
- **H₁:** existe asociación. **α = 0,05.**

Requiere NumPy y SciPy; no necesita archivos de datos.

```python
import numpy as np
from scipy import stats

tabla = np.array([[30, 70], [50, 50]])
# Pearson sin corrección de continuidad; esperadas suficientemente grandes.
chi2, p, gl, esperadas = stats.chi2_contingency(tabla, correction=False)
print('Frecuencias esperadas:\n', esperadas)
assert (esperadas >= 5).all()
print(f'Chi-cuadrado = {chi2:.2f}; gl = {gl}; p = {p:.4f}')
print('Rechazamos H₀.' if p < 0.05 else 'No rechazamos H₀.')
```

## 3. Interpretación

En la muestra compró el **30 %** de los clientes de móvil y el **50 %** de los de ordenador. Como **p = 0.0039 < 0,05**, hay evidencia de asociación.

Esto no demuestra que el dispositivo cause la compra.
