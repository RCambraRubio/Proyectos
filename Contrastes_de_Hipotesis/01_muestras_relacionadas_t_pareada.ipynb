# Muestras relacionadas: t pareada

[Fundamentos](00_fundamentos.ipynb) · [Muestras independientes](03_muestras_independientes_t_student.ipynb)

La **t pareada** evalúa si el **cambio medio** entre dos mediciones relacionadas es distinto de cero. Se calcula una diferencia por persona: **D = después − antes**.

## 1. Supuestos

| Supuesto | Qué significa |
|---|---|
| **Parejas bien definidas** | Cada «antes» corresponde al «después» de la misma persona. |
| **Datos cuantitativos comparables** | Ambas medidas usan la misma escala y sus diferencias tienen sentido. |
| **Independencia entre personas** | El cambio de una persona no depende del de otra. |
| **Normalidad de las diferencias** | Con muestras pequeñas, las diferencias deben ser aproximadamente normales. Con muestras grandes, puede bastar una aproximación por TCL, con varianza finita y sin extremos dominantes. |

No exige normalidad de «antes» y «después» por separado, ni igualdad de sus varianzas.


## 2. Ejemplo: puntuación antes y después

Usamos los datos ficticios de **36 personas** de `datos/relacionadas.json`. Para el ejemplo, suponemos personas independientes y cambios aproximadamente normales; esto se debe revisar en un análisis real.

- **H₀:** cambio medio = 0.
- **H₁:** cambio medio ≠ 0 (bilateral).
- **α = 0,05.** Un cambio positivo indica mayor puntuación después.

Ejecuta desde la carpeta del notebook; requiere pandas y SciPy.


```python
import pandas as pd
from scipy import stats

df = pd.read_json('datos/relacionadas.json')
assert df['id'].is_unique and df[['antes', 'despues']].notna().all().all()
df['cambio'] = df['despues'] - df['antes']
print(df.head().to_string(index=False))

resultado = stats.ttest_rel(df['despues'], df['antes'])
ic = resultado.confidence_interval(confidence_level=0.95)
print(f'\nPersonas: {len(df)}; cambio medio: {df.cambio.mean():.2f} puntos')
print(f'IC 95 %: [{ic.low:.2f}, {ic.high:.2f}] puntos')
print(f't = {resultado.statistic:.2f}; p = {resultado.pvalue:.4f}')
print('Rechazamos H₀.' if resultado.pvalue < 0.05 else 'No rechazamos H₀.')
```

## 3. Interpretación

La puntuación aumentó **3.36 puntos de media** (IC 95 %: **1.26 a 5.47**). Como **p = 0.0026 < 0,05**, rechazamos H₀: hay evidencia de un cambio medio positivo, bajo los supuestos.

El antes/después por sí solo no demuestra qué causó el cambio.
